```python
total_revenue = df["Revenue_USD"].sum()
print("Total Revenue_USD:", total_revenue)
```

```python
total_orders = df["Sale_ID"].nunique()
print("Total Orders:", total_orders)
```

```python
total_units = df["No_of_Units_Sold"].sum()
print("Total Units Sold:", total_units)
```

```python
average_order_value = df["Revenue_USD"].sum() / df["Sale_ID"].nunique()
print("Average Order Value:", average_order_value)
```

```python
revenue_by_year = df.groupby("Year")["Revenue_USD"].sum().sort_index()
print(revenue_by_year)
```

```python
revenue_by_quarter = (
    df.groupby(["Year", "Quarter"])["Revenue_USD"]
      .sum()
      .reset_index()
)

print(revenue_by_quarter)
```

```python
monthly_revenue = (
    df.groupby(["Year", "Month_of_Sale"])["Revenue_USD"]
      .sum()
      .reset_index()
)

print(monthly_revenue)
```

```python
revenue_by_country = (
    df.groupby("Country")["Revenue_USD"]
      .sum()
      .sort_values(ascending=False)
)

print(revenue_by_country)
```

```python
top_10_countries = (
    df.groupby("Country")["Revenue_USD"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_10_countries)
```

```python
orders_by_country = (
    df.groupby("Country")["Sale_ID"]
      .nunique()
      .sort_values(ascending=False)
)

print(orders_by_country)
```

```python
aov_by_country = (
    df.groupby("Country")
      .apply(lambda x: x["Revenue_USD"].sum() / x["Sale_ID"].nunique())
      .sort_values(ascending=False)
)

print(aov_by_country)
```

```python
revenue_by_category = (
    df.groupby("Category")["Revenue_USD"]
      .sum()
      .sort_values(ascending=False)
)

print(revenue_by_category)
```

```python
units_by_category = (
    df.groupby("Category")["No_of_Units_Sold"]
      .sum()
      .sort_values(ascending=False)
)

print(units_by_category)
```

```python
revenue_by_channel = (
    df.groupby("Sales_Channel")["Revenue_USD"]
      .sum()
      .sort_values(ascending=False)
)

print(revenue_by_channel)
```

```python
revenue_by_segment = (
    df.groupby("Customer_Segment")["Revenue_USD"]
      .sum()
      .sort_values(ascending=False)
)

print(revenue_by_segment)
```

```python
revenue_by_payment = (
    df.groupby("Payment_Method")["Revenue_USD"]
      .sum()
      .sort_values(ascending=False)
)

print(revenue_by_payment)
```

```python
revenue_by_age_group = (
  df.groupby("Customer_age_group")["Revenue_USD"]
    .sum()
    .sort_values(ascending=False)
)

print(revenue_by_age_group)

print(revenue_by_age_group)
```

```python
return_analysis = (
    df.groupby("Return_Status")
      .agg(
          Total_Orders=("Sale_ID", "nunique"),
          Total_Revenue=("Revenue_USD", "sum")
      )
      .reset_index()
)

print(return_analysis)
```

```python
country_category_revenue = (
    df.groupby(["Country", "Category"])["Revenue_USD"]
      .sum()
      .reset_index()
      .sort_values(["Country", "Revenue_USD"], ascending=[True, False])
)

print(country_category_revenue)
```

```python
year_category_revenue = (
    df.groupby(["Year", "Category"])["Revenue_USD"]
      .sum()
      .reset_index()
)

top_category_each_year = (
    year_category_revenue.loc[
        year_category_revenue.groupby("Year")["Revenue_USD"].idxmax()
    ]
)

print(top_category_each_year)
```

```python
plt.figure(figsize=(10, 6))

revenue_by_year.plot(kind="line", marker="o")

plt.title("Revenue Trend by Year")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

top_10_countries.sort_values().plot(kind="barh")

plt.title("Top 10 Countries by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Country")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

revenue_by_category.plot(kind="bar")

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

units_by_category.plot(kind="bar")

plt.title("Units Sold by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Units Sold")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 6))

revenue_by_segment.plot(kind="bar")

plt.title("Revenue by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 6))

revenue_by_channel.plot(kind="bar")

plt.title("Revenue by Sales Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```
